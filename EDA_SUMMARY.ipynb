## EDA Pipeline 

The exploratory data analysis was conducted in a structured manner:

1. **Data Ingestion**
   - Raw CSV files were loaded into a SQLite database without modification
   - Schema and row counts were validated

2. **SQL Sanity Checks**
   - Primary and foreign key integrity was verified
   - No orphaned records or invalid values were found

3. **Univariate Analysis**
   - User activity, product popularity, reorder ratios, and purchase cadence
     were analyzed individually
   - All major distributions were found to be highly skewed and long-tailed

4. **Bivariate Analysis**
   - Relationships between users, products, and contextual variables were examined
   - Strong interaction effects were observed

5. **Correlation and Multicollinearity**
   - High correlation among count-based features was identified
   - Derived ratios provided complementary behavioral signals


## Key Behavioral Insights

Several consistent patterns emerged from EDA:

- **User behavior is highly heterogeneous**
  - A small number of users place many orders
  - Most users are relatively inactive

- **Product demand follows a long-tail distribution**
  - Few staple products dominate overall purchases
  - Many products are niche or infrequently purchased

- **Reorder behavior is uneven**
  - Some products and users exhibit strong habitual behavior
  - Others show exploratory or one-time purchase patterns

These findings indicate that reorder behavior cannot be explained
by single features or simple rules.


## Why User–Product Level Modeling Is Necessary

EDA showed that:
- User-level features alone are too coarse
- Product-level features ignore personalization
- Contextual features influence behavior but are insufficient on their own

The strongest signal for reorder behavior emerges at the
**user–product interaction level**, where we observe:
- How frequently a user buys a product
- How consistently the user reorders that product

This motivates framing the problem as:
> Predicting whether a given **user–product pair** will be reordered.


## Feature Types Identified from EDA

Based on exploratory analysis, features naturally fall into the following groups:

### User-Level Features
- Total number of orders
- Purchase cadence (days since prior order)

### Product-Level Features
- Total purchase count
- Product reorder ratio
- Department and aisle context

### User–Product Interaction Features
- Times a user bought a product
- Times a user reordered a product
- User–product reorder ratio

EDA indicates that interaction features are the most informative.


## Modeling Implications

Several modeling considerations follow directly from EDA:

- Feature distributions are **non-Gaussian** and highly skewed
- Relationships between variables are **non-linear**
- Feature interactions play a critical role
- Multicollinearity is present but structurally meaningful

As a result:
- Simple linear models may struggle
- Tree-based and ensemble models are better suited
- Model interpretability should be addressed post hoc
